```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Tiny synthetic healthcare dataset
# Feature: Age (years)
# Target: Blood Pressure (mmHg)

data = pd.DataFrame({
    "age": [20, 30, 35, 40, 45, 50, 55, 60],
    "systolic_bp": [112, 118, 121, 127, 130, 136, 140, 146]
})

data
```

```python
data.describe()
```

```python
data.shape
```

```python
X = data["age"].values
y = data["systolic_bp"].values

print("X:", X)
print("y:", y)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    data["age"],
    data["systolic_bp"],
    s=80
)

plt.xlabel("Age (years)")
plt.ylabel("Systolic Blood Pressure (mmHg)")
plt.title("Age vs Systolic Blood Pressure")

plt.grid(True, alpha=0.3)
plt.show()
```

```python
x_mean = np.mean(X)
y_mean = np.mean(y)

print("Mean Age:", x_mean)
print("Mean Systolic BP:", y_mean)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(X, y, s=80)

plt.axvline(x_mean, linestyle="--", label=f"Mean Age = {x_mean:.1f}")
plt.axhline(y_mean, linestyle="--", label=f"Mean BP = {y_mean:.1f}")

plt.xlabel("Age (years)")
plt.ylabel("Systolic Blood Pressure (mmHg)")
plt.title("Center of the Healthcare Data")

plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
